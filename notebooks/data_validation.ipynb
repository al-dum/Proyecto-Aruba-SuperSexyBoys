# Notebook para la validación de los datos de las lecturas "data/raw/aruba_readings_20260924.json" mediante ETL

```python
import pandas as pd
json_data_values = pd.read_json("/home/igomez/WSPython/Proyecto-Aruba-SuperSexyBoys/data/raw/readings/aruba_readings_20260924.json")
json_data_stations_ordered = pd.read_json("/home/igomez/WSPython/Proyecto-Aruba-SuperSexyBoys/data/processed/aruba_stations_ordered_by_distance_m.json")
```

```python
json_data_values = json_data_values[
    json_data_values["station_id"].isin(
        json_data_stations_ordered["id"]
    )
]

json_data_values = json_data_values.sort_values(by="timestamp", ascending=True).reset_index(drop=True)

json_data_values.to_csv("/home/igomez/WSPython/Proyecto-Aruba-SuperSexyBoys/data/processed/aruba_reading_filtered_by_station_distance.csv", index=False)
json_data_values.to_json("/home/igomez/WSPython/Proyecto-Aruba-SuperSexyBoys/data/processed/aruba_reading_filtered_by_station_distance.json")
```
