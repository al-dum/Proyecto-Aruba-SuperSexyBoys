# Notebook para la validación de los datos de las lecturas "data/raw/aruba_readings_20260924.json" mediante ETL

```python
import pandas as pd
```

Importamos los json de lecturas y las estaciones ordenadas por distancia.

```python
json_data_values = pd.read_json("../data/raw/readings/aruba_readings_20260924.json")
json_data_stations_ordered = pd.read_json("/home/igomez/WSPython/Proyecto-Aruba-SuperSexyBoys/data/processed/aruba_stations_ordered_by_distance_m.json")
```

Se filtran las lecturas en función del identificador de la estación para aislar el volumen de datos objetivo

```python
json_station_data_values = json_data_values[
    json_data_values["station_id"].isin(
        json_data_stations_ordered["id"]
    )
]

json_station_data_values = json_data_values.sort_values(by="timestamp", ascending=True).reset_index(drop=True)

json_station_data_values.head()
```

Exportamos los datos filtrados por estación a csv o json.

```python
json_station_data_values.to_csv("/home/igomez/WSPython/Proyecto-Aruba-SuperSexyBoys/data/processed/aruba_reading_filtered_by_station_distance.csv", index=False)
json_station_data_values.to_json("/home/igomez/WSPython/Proyecto-Aruba-SuperSexyBoys/data/processed/aruba_reading_filtered_by_station_distance.json")
```
