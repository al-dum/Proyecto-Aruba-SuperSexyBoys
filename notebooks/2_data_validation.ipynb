# Notebook para la validación de las lecturas limpias mediante ETL

```python
import pandas as pd
```

Importamos las lecturas limpias del paso 0 y las estaciones ordenadas por distancia del paso 1.

```python
json_data_values = pd.read_json("../data/processed/0_cleaning/aruba_readings_cleaned.json")
json_data_stations_ordered = pd.read_json("../data/processed/1_station_filtering/aruba_stations_ordered_by_distance_m.json")
```

Se filtran las lecturas en función del identificador de la estación para aislar el volumen de datos objetivo

```python
json_station_data_values = json_data_values[
    json_data_values["station_id"].isin(
        json_data_stations_ordered["id"]
    )
]

json_station_data_values = json_data_values.sort_values(by="timestamp", ascending=True).reset_index(drop=True)

json_station_data_values.head()
```

Exportamos los datos filtrados por estación a csv o json.

```python
json_station_data_values.to_csv("../data/processed/2_data_validation/aruba_reading_filtered_by_station_distance.csv", index=False)
json_station_data_values.to_json("../data/processed/2_data_validation/aruba_reading_filtered_by_station_distance.json")
```
